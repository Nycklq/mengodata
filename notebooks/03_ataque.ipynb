# Produção ofensiva

Gols, assistências, G+A e taxas por 90 exigem observações por partida e minutos confiáveis. Ausências permanecem nulas. Totais anuais só são interpretados com cobertura integral auditada.

```python
from pathlib import Path
from flamengo_pipeline.infrastructure.research_files import read_research
from flamengo_pipeline.application.research_analysis import analyze_players, analyze_metric

root = Path.cwd()
if not (root / 'contracts/scope.json').exists():
    root = root.parent
state = read_research(root)
players = analyze_players(state['document']) if state['document'] else []
print({'disponibilidade': state['availability'], 'motivo': state['reason']})
```

## Verificação reproduzível

A saída abaixo é produzida pelos módulos do projeto. Fixtures sintéticas de testes não são carregadas.

```python
if not players:
    print('Análise ofensiva bloqueada: não há observações individuais autorizadas.')
else:
    known = [p for p in players if p['contributions_per90'] is not None]
    print(sorted(known, key=lambda p: p['contributions_per90'], reverse=True))
```
