# Comparação entre atletas

A comparação utiliza o mesmo recorte, mínimo de minutos e unidade de medida. Jogadores de posições diferentes exercem funções distintas; não se infere superioridade universal a partir de um indicador.

```python
from pathlib import Path
from flamengo_pipeline.infrastructure.research_files import read_research
from flamengo_pipeline.application.research_analysis import analyze_players, analyze_metric

root = Path.cwd()
if not (root / 'contracts/scope.json').exists():
    root = root.parent
state = read_research(root)
players = analyze_players(state['document']) if state['document'] else []
print({'disponibilidade': state['availability'], 'motivo': state['reason']})
```

## Verificação reproduzível

A saída abaixo é produzida pelos módulos do projeto. Fixtures sintéticas de testes não são carregadas.

```python
if not players:
    print('Comparação bloqueada: seleções reais dependem de identidades e aparições verificadas.')
else:
    eligible = [p for p in players if p['minutes'] is not None and p['minutes'] >= 900]
    print({'limiar_minutos': 900, 'jogadores_elegiveis': [p['player_id'] for p in eligible]})
```
